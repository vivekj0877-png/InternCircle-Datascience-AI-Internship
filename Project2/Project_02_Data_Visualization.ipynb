```python

# Import required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully!")
```

```python

# Load Tips dataset

df = sns.load_dataset("tips")

print("Tips dataset loaded successfully!")
```

```python

# Display first five rows

df.head()
```

```python

print("Dataset Shape:", df.shape)
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])
```

```python

df.info()
```

```python

df.describe()
```

```python

print("Missing Values:")
print(df.isnull().sum())
```

```python

print("Duplicate Records:", df.duplicated().sum())
```

```python

# Set visualization theme

sns.set_theme(style="whitegrid", palette="deep")

print("Visualization theme applied!")
```

```python

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="total_bill",
    bins=20,
    color="steelblue"
)

plt.title("Distribution of Restaurant Bills")
plt.xlabel("Total Bill (USD)")
plt.ylabel("Number of Meals")

plt.show()
```

```python

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="day",
    y="tip",
    palette="deep",
    hue="day",
    legend=False
)

plt.title("Distribution of Tips by Day")
plt.xlabel("Day")
plt.ylabel("Tip Amount (USD)")

plt.show()
```

```python

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="time",
    palette="deep"
)

plt.title("Relationship Between Total Bill and Tip")
plt.xlabel("Total Bill (USD)")
plt.ylabel("Tip Amount (USD)")

plt.show()
```

```python

# Select numerical columns

numeric_data = df.select_dtypes("number")

# Calculate correlation

correlation = numeric_data.corr()

# Create heatmap

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="Blues",
    fmt=".2f"
)

plt.title("Correlation Between Numerical Variables")

plt.show()
```

```python

# Create a 2x2 dashboard

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Chart 1: Histogram
sns.histplot(
    df["total_bill"],
    bins=20,
    ax=axes[0, 0]
)

axes[0, 0].set(
    title="Distribution of Restaurant Bills",
    xlabel="Total Bill (USD)",
    ylabel="Number of Meals"
)

# Chart 2: Box Plot
sns.boxplot(
    data=df,
    x="day",
    y="tip",
    hue="day",
    palette="deep",
    legend=False,
    ax=axes[0, 1]
)

axes[0, 1].set(
    title="Tips by Day",
    xlabel="Day",
    ylabel="Tip (USD)"
)

# Chart 3: Scatter Plot
sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="time",
    ax=axes[1, 0]
)

axes[1, 0].set(
    title="Relationship Between Bill and Tip",
    xlabel="Total Bill (USD)",
    ylabel="Tip (USD)"
)

# Chart 4: Heatmap
sns.heatmap(
    df.select_dtypes("number").corr(),
    annot=True,
    cmap="Blues",
    fmt=".2f",
    ax=axes[1, 1]
)

axes[1, 1].set_title("Correlation Between Numbers")

# Adjust layout
plt.tight_layout()

# Save dashboard
plt.savefig("tips_overview.png", dpi=150, bbox_inches="tight")

plt.show()

print("Dashboard created and saved successfully!")
```

CHART 1 – HISTOGRAM
Shows the distribution of restaurant bills.

CHART 2 – BOX PLOT
Compares tip distributions across different days.

CHART 3 – SCATTER PLOT
Shows the relationship between total bill and tip.

CHART 4 – CORRELATION HEATMAP
Shows the relationships between numerical variables.
